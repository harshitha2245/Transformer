```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")

classifier("I love learning Transformers!")
```

```python
model = pipeline("fill-mask", model="google-bert/bert-base-uncased")

model("Paris is the [MASK] of France.")
```

```python
generator = pipeline("text-generation", model="openai-community/gpt2")

generator("Artificial Intelligence is", max_new_tokens=30)
```

```python
# In newer transformers versions (v5+), legacy tasks like 'text2text-generation' are removed.
# We must use 'text-generation' instead.
qa_generator = pipeline("text-generation", model="openai-community/gpt2")

context = "Transformers were introduced by researchers at Google in 2017."
question = "Who developed Transformers?"

# Constructing a prompt for the model
prompt = f"Context: {context}\nQuestion: {question}\nAnswer:"
qa_generator(prompt, max_new_tokens=20)
```

```python
ner = pipeline("ner", model="dslim/bert-base-NER")

ner("Rahul works at Google in Mumbai.")
```

```python
# In transformers v5+, 'text2text-generation' is removed. We use 'text-generation' with a general model instead.
translator = pipeline("text-generation", model="openai-community/gpt2")

prompt = "Translate English to German: I love machine learning. Translation:"
translator(prompt, max_new_tokens=20)
```
